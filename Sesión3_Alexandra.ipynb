**Clasificador de notas**

```python
def estado_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota invalida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    return "Excelente"

notas = [2.5, 3.8, 4.7, 5.2, 3.0]
aprobados = 0
for n in notas:
    estado = estado_nota(n)
    print(n, "-", estado)
    if estado in ("Aprueba", "Excelente"):
        aprobados += 1
print("Aprobados:", aprobados)
```

**Carrito de compras**

```python
# Carrito: producto -> (precio, cantidad)
carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}

# Calcular subtotal
subtotal = 0

for producto, (precio, cantidad) in carrito.items():
    subtotal += precio * cantidad

print("Subtotal: $", format(subtotal, ","))

# Aplicar descuento si supera 200.000
if subtotal > 200000:
    descuento = subtotal * 0.10
else:
    descuento = 0

subtotal_descuento = subtotal - descuento

# Aplicar IVA del 19%
iva = subtotal_descuento * 0.19
total = subtotal_descuento + iva

print("Descuento: $", format(int(descuento), ","))
print("IVA: $", format(int(iva), ","))
print("Total a pagar: $", format(int(total), ","))
```

**Nómina con pandas**

```python
import pandas as pd

# DataFrame inicial
empleados = pd.DataFrame({
    "nombre": ["Ana", "Carlos", "Laura"],
    "area": ["Ventas", "RRHH", "Tecnología"],
    "salario": [3000000, 2500000, 4000000]
})

# Agregar 3 empleados nuevos
nuevos = pd.DataFrame({
    "nombre": ["Brandon", "Juliana", "Andrés"],
    "area": ["Ventas", "Contabilidad", "Ventas"],
    "salario": [3500000, 2800000, 4500000]
})

empleados = pd.concat([empleados, nuevos], ignore_index=True)

# Crear columna bono
empleados["bono"] = empleados.apply(
    lambda fila: fila["salario"] * 0.10
    if fila["area"] == "Ventas"
    else fila["salario"] * 0.05,
    axis=1
)

# Costo total (salario + bono)
empleados["costo_total"] = empleados["salario"] + empleados["bono"]

print("Tabla de empleados:")
print(empleados)

# Nómina total por área
nomina_area = empleados.groupby("area")["costo_total"].sum()

print("\nNómina mensual por área:")
print(nomina_area)

# Empleado con mayor salario
mayor_salario = empleados.loc[empleados["salario"].idxmax()]

print("\nEmpleado con mayor salario:")
print(mayor_salario["nombre"])
print("$", format(int(mayor_salario["salario"]), ","))
```
