**Revisa tu ticket de compra**

```python
# Precios de tu compra
precios = [1000, 900, 500, 700]

# 1. Imprime cada precio con un for
for precio in precios:
    print(precio)

# 2. Suma el total
total = 0

for precio in precios:
    total += precio

print("Total:", total)

# 3. ¿Cuántos cuestan más de 100?
caros = 0

for precio in precios:
    if precio > 100:
        caros += 1

print("Productos mayores a 100:", caros)
```

**Tu meta de ahorro**

```python
ahorro = 0
meta = 15000000
por_mes = 1500000
meses = 0

# Mientras no llegues a la meta
while ahorro < meta:
    ahorro += por_mes
    meses += 1
    print("Mes", meses, "- Ahorro acumulado:", ahorro)

# ¿Cuántos meses tardaste?
print("Meses para alcanzar la meta:", meses)
```

**Dividir la cuenta**

```python
cuenta = 177300
propina = 10
amigos = 3

def con_propina(cuenta, porcentaje):
    return cuenta + (cuenta * porcentaje / 100)

def por_persona(total, personas):
    return total / personas

def descuento(total, porcentaje):
    return total - (total * porcentaje / 100)

total = con_propina(cuenta, propina)

print("Total con propina:", total)
print("Cada persona paga:", por_persona(total, amigos))
print("Total con 15% de descuento:", descuento(cuenta, 15))
print("Cada persona paga:", por_persona(descuento(cuenta, 10), amigos))
```

**La caja de la tienda**

```python
# Lista de precios
precios = [10000, 30000, 40000, 60000]

# 1. Sumar el total con un for
total = 0

for precio in precios:
    total += precio

# 2. Aplicar descuento si supera 500
if total > 500:
    descuento = total * 0.10
else:
    descuento = 0

a_pagar = total - descuento

# 3. Imprimir ticket
print(f"Total: ${total}")
print(f"Descuento: ${descuento}")
print(f"A pagar: ${a_pagar}")
```
